```python
# ============================================================
# EXPLORATORY DATA ANALYSIS (EDA) PROJECT
# Titanic Dataset
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set_style("whitegrid")


# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv('/content/titanic_eda_dataset.csv')

print("Dataset loaded successfully!")


# ============================================================
# 3. UNDERSTAND THE DATASET
# ============================================================

print("\n========== FIRST 5 ROWS ==========")
display(df.head())

print("\n========== LAST 5 ROWS ==========")
display(df.tail())

print("\n========== DATASET SHAPE ==========")
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\n========== COLUMN NAMES ==========")
print(df.columns.tolist())

print("\n========== DATASET INFORMATION ==========")
df.info()


# ============================================================
# 4. STATISTICAL SUMMARY
# ============================================================

print("\n========== STATISTICAL SUMMARY ==========")
display(df.describe())


# ============================================================
# 5. MISSING VALUES
# ============================================================

print("\n========== MISSING VALUES ==========")

missing_values = df.isnull().sum()

display(missing_values)

print("Total missing values:", missing_values.sum())


# ============================================================
# 6. DUPLICATE VALUES
# ============================================================

print("\n========== DUPLICATE ROWS ==========")

duplicates = df.duplicated().sum()

print("Number of duplicate rows:", duplicates)


# ============================================================
# 7. UNIQUE VALUES
# ============================================================

print("\n========== UNIQUE VALUES ==========")

for column in df.columns:
    print(column, ":", df[column].nunique())


# ============================================================
# 8. NUMERICAL COLUMNS
# ============================================================

numeric_columns = df.select_dtypes(include='number').columns

print("\n========== NUMERICAL COLUMNS ==========")
print(numeric_columns.tolist())


# ============================================================
# 9. HISTOGRAMS
# ============================================================

print("\n========== HISTOGRAMS ==========")

df[numeric_columns].hist(
    figsize=(12, 8),
    bins=20
)

plt.suptitle("Distribution of Numerical Variables")
plt.tight_layout()
plt.show()


# ============================================================
# 10. BOXPLOTS
# ============================================================

print("\n========== BOXPLOTS ==========")

for column in numeric_columns:

    plt.figure(figsize=(6, 4))

    sns.boxplot(x=df[column])

    plt.title("Box Plot of " + column)
    plt.xlabel(column)

    plt.show()


# ============================================================
# 11. CORRELATION ANALYSIS
# ============================================================

print("\n========== CORRELATION MATRIX ==========")

correlation_matrix = df[numeric_columns].corr()

display(correlation_matrix)


# ============================================================
# 12. CORRELATION HEATMAP
# ============================================================

print("\n========== CORRELATION HEATMAP ==========")

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()


# ============================================================
# 13. SURVIVAL ANALYSIS
# ============================================================

print("\n========== SURVIVAL COUNT ==========")

survival_count = df["Survived"].value_counts()

print(survival_count)

plt.figure(figsize=(6, 4))

sns.countplot(
    data=df,
    x="Survived"
)

plt.title("Titanic Survival Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")

plt.show()


# ============================================================
# 14. OVERALL SURVIVAL RATE
# ============================================================

overall_survival_rate = df["Survived"].mean() * 100

print("\n========== OVERALL SURVIVAL RATE ==========")

print(
    f"Overall survival rate: {overall_survival_rate:.2f}%"
)


# ============================================================
# 15. SURVIVAL BY GENDER
# ============================================================

print("\n========== SURVIVAL BY GENDER ==========")

gender_table = pd.crosstab(
    df["Sex"],
    df["Survived"]
)

display(gender_table)


# Gender survival rate

gender_survival = (
    df.groupby("Sex")["Survived"]
    .mean() * 100
)

print("\nSurvival Rate by Gender:")

for gender, rate in gender_survival.items():

    print(
        f"{gender}: {rate:.2f}%"
    )


# Visualization

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Sex",
    hue="Survived"
)

plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.show()


# ============================================================
# 16. SURVIVAL BY PASSENGER CLASS
# ============================================================

print("\n========== SURVIVAL BY PASSENGER CLASS ==========")

class_table = pd.crosstab(
    df["Pclass"],
    df["Survived"]
)

display(class_table)


# Class survival rate

class_survival = (
    df.groupby("Pclass")["Survived"]
    .mean() * 100
)

print("\nSurvival Rate by Passenger Class:")

for passenger_class, rate in class_survival.items():

    print(
        f"Class {passenger_class}: {rate:.2f}%"
    )


# Visualization

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Pclass",
    hue="Survived"
)

plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.show()


# ============================================================
# 17. AGE ANALYSIS
# ============================================================

print("\n========== AGE ANALYSIS ==========")

age_by_survival = (
    df.groupby("Survived")["Age"]
    .mean()
)

print(
    f"Average age of survivors: "
    f"{age_by_survival.get(1, 0):.2f} years"
)

print(
    f"Average age of non-survivors: "
    f"{age_by_survival.get(0, 0):.2f} years"
)


# Age distribution

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Age",
    hue="Survived",
    bins=20,
    kde=True
)

plt.title("Age Distribution by Survival")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()


# ============================================================
# 18. FARE ANALYSIS
# ============================================================

print("\n========== FARE ANALYSIS ==========")

fare_by_survival = (
    df.groupby("Survived")["Fare"]
    .mean()
)

print(
    f"Average fare of survivors: "
    f"{fare_by_survival.get(1, 0):.2f}"
)

print(
    f"Average fare of non-survivors: "
    f"{fare_by_survival.get(0, 0):.2f}"
)


# Fare distribution

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Survived",
    y="Fare"
)

plt.title("Fare Distribution by Survival")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Fare")

plt.show()


# ============================================================
# 19. KEY INFLUENCING FACTORS
# ============================================================

print("\n========== KEY INFLUENCING FACTORS ==========")

highest_gender = gender_survival.idxmax()
highest_gender_rate = gender_survival.max()

highest_class = class_survival.idxmax()
highest_class_rate = class_survival.max()

print(
    f"Gender with highest survival rate: "
    f"{highest_gender} ({highest_gender_rate:.2f}%)"
)

print(
    f"Passenger class with highest survival rate: "
    f"Class {highest_class} ({highest_class_rate:.2f}%)"
)


# ============================================================
# 20. DATA-BASED INSIGHTS
# ============================================================

print("\n================================================")
print("              KEY EDA INSIGHTS")
print("================================================")

print(
    f"\n1. The Titanic dataset contains "
    f"{df.shape[0]} passengers and "
    f"{df.shape[1]} columns."
)

print(
    f"\n2. The overall survival rate was "
    f"{overall_survival_rate:.2f}%."
)

print(
    f"\n3. {highest_gender} passengers had the highest "
    f"survival rate at {highest_gender_rate:.2f}%."
)

print(
    f"\n4. Passenger Class {highest_class} had the highest "
    f"survival rate at {highest_class_rate:.2f}%."
)

print(
    f"\n5. The average age of survivors was "
    f"{age_by_survival.get(1, 0):.2f} years, while "
    f"non-survivors had an average age of "
    f"{age_by_survival.get(0, 0):.2f} years."
)

print(
    f"\n6. The average fare paid by survivors was "
    f"{fare_by_survival.get(1, 0):.2f}, while "
    f"non-survivors paid an average fare of "
    f"{fare_by_survival.get(0, 0):.2f}."
)


# ============================================================
# 21. FINAL CONCLUSION
# ============================================================

print("\n================================================")
print("                 CONCLUSION")
print("================================================")

print("""
The Exploratory Data Analysis of the Titanic dataset
was performed using statistical summaries and visualizations.

The analysis identified patterns and trends in passenger
survival. Gender and passenger class showed noticeable
differences in survival rates. Age and fare also showed
differences between survivors and non-survivors.

Correlation analysis was used to understand relationships
between numerical variables. Histograms, box plots,
count plots and a correlation heatmap were used to
visually explore the dataset.

Overall, the analysis demonstrates how statistical
analysis and data visualization can be used to identify
important patterns, relationships and influencing factors
in a real-world dataset.
""")


# ============================================================
# PROJECT COMPLETED
# ============================================================

print("\n================================================")
print("       EXPLORATORY DATA ANALYSIS COMPLETED")
print("================================================")
```
